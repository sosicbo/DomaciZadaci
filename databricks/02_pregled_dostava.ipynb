# Pregled dostava po gradu

Notebook broji dostave po gradu. Grad (`region`) se dobija spajanjem tabele `dostave` sa tabelom `zone_dostave`,
preko zone magacina iz koje je paket preuzet (`pickup_location_id = location_id`).

Isti rezultat se računa u SQL ćeliji i u Python ćeliji (preko `spark.sql` i `toPandas`), da bi se proverilo da se poklapaju.

## Šta se vidi na prikazima

- **Dostave po gradu:** koji grad ima najviše, a koji najmanje dostava.
- **Broj paketa po dostavi:** prikaz koliko paketa najcešce ide u jednoj dostavi.
- **Dostave po satu u danu:** prikaz u koje doba dana ima najvise posla.

```python
%sql
--Broj dostava po gradu
SELECT 
zd.region AS grad,
COUNT(*) AS broj_dostava
FROM workspace.source.dostave d
JOIN workspace.source.zone_dostave zd ON d.pickup_location_id=zd.location_id
GROUP BY zd.region
ORDER BY broj_dostava DESC
```

```python
%sql
--Zbir po gradovima
SELECT COUNT(*)
FROM workspace.source.dostave d
JOIN workspace.source.zone_dostave zd ON d.pickup_location_id=zd.location_id
```

```python
# spark.sql i toPandas
query="""
SELECT 
zd.region AS grad,
COUNT(*) AS broj_dostava
FROM workspace.source.dostave d
JOIN workspace.source.zone_dostave zd ON d.pickup_location_id=zd.location_id
GROUP BY zd.region
ORDER BY broj_dostava DESC
"""

dostave_po_gradu=spark.sql(query).toPandas()

print(dostave_po_gradu)
print("Zbir po gradovima:",dostave_po_gradu['broj_dostava'].sum())
```

```python
#Dostave po gradu
display(spark.sql(query))
```

```python
#Broj paketa po dostavi
query="""
SELECT package_count AS br_paketa
, COUNT(*) AS br_dostava
FROM workspace.source.dostave
GROUP BY package_count
ORDER BY br_paketa
"""
display(spark.sql(query))
```

```python
#Dostave po satu u danu
query="""
SELECT 
HOUR(pickup_datetime) AS sat,
COUNT(*) as br_dostava
FROM workspace.source.dostave
GROUP BY HOUR(pickup_datetime)
ORDER BY sat
"""
display(spark.sql(query))
```
